```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv")
```

```python
df = df.dropna(subset="Age")
```

```python
fig, axs = plt.subplots(2, 2, figsize=(10, 8))

axs[0, 0].hist(df["Age"], color="white", edgecolor="blue")

axs[0, 1].bar(df["Pclass"].value_counts().index, df["Pclass"].value_counts(), color="black", edgecolor="blue")

axs[1, 0].scatter(df["Age"], df["Fare"])

axs[1, 1].plot(df.groupby('Age')['Fare'].mean())

print(df["Pclass"].value_counts())
```
