```python
import numpy as np
import matplotlib.pyplot as plt
import random
x=np.linspace(0,100,100000)
y = np.sin(x) + np.cos(x)+x
y2 = random.randint(-9,10)*x
y3 = random.randint(1,1000) + x

fig, axs = plt.subplots(1,3,figsize = (12,5))
axs[0].plot(x,y)
axs[0].plot(x, y2)
axs[0].plot(x, y3)
axs[1].plot(x,y2)
axs[2].plot(x,y3)

plt.show()
```

```python

datos = np.random.randn(1000)  # Genera 1000 puntos con distribución normal

plt.hist(datos, bins=10, color='green', alpha=0.7)  # Dibuja el histograma
plt.title("Histograma de Datos Aleatorios")  # Título
plt.xlabel("Valor")  # Etiqueta del eje x
plt.ylabel("Frecuencia")  # Etiqueta del eje y
plt.show()  # Muestra el gráfico
```
