```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv")
```

```python
print(df.head(5))
```

```python
df.info()
```

```python
lista_edades = df['Age']
```

```python
edad_promedio = lista_edades.mean()
print(edad_promedio)
```

```python
gente_premium = df[df['Fare'] > 100]
print(gente_premium)
print(len(gente_premium))
```

```python
print(len(gente_premium))
```

```python
print(df.isnull().sum())
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
print(df.isnull().sum())
```

```python
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
```

```python
print(df)
```

```python
promedio_de_supervivencia = df.groupby('Sex')['Survived'].mean() * 100
print(promedio_de_supervivencia)
```

```python
precio_prom = df['Fare'].mean()
precio_mas_alto = df.loc[df['Fare'].idxmax()]
precio_mas_bajo = df.loc[df['Fare'].idxmin()]
print(f"Precio promedio: {precio_prom}")
print(f"Precio más caro: {precio_mas_alto['Fare']}")
print(f"Precio más barato: {precio_mas_bajo['Fare']}")
```
