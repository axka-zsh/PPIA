```python
import re

CURP_REGEX = re.compile(
    r"^[A-Z]{4}"          # 4 letras iniciales
    r"\d{6}"              # fecha de nacimiento AAMMDD
    r"[HM]"               # sexo
    r"[A-Z]{2}"           # entidad de nacimiento
    r"[B-DF-HJ-NP-TV-Z]{3}"  # 3 consonantes (sin vocales)
    r"[A-Z0-9]"           # diferenciador de homoclave
    r"\d$"                # digito verificador
)


def es_curp_valido(curp: str) -> bool:
    return bool(CURP_REGEX.match(curp.strip().upper()))

curp_input = input("Ingresa CURP a comprobar")
print("CURP válido" if es_curp_valido(curp_input) else "CURP inválido")
```
